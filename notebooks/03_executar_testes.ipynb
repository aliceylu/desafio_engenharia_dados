```python
import pytest

print(f"Versão do pytest: {pytest.__version__}")
```

```python
import os
import sys
from pathlib import Path

diretorio = Path(os.getcwd()).resolve()

repo = next(
    (
        p for p in (diretorio, *diretorio.parents)
        if (p / "main.py").is_file()
        and (p / "tests" / "conftest.py").is_file()
    ),
    None
)

if repo is None:
    raise FileNotFoundError(
        "Não foi possível localizar a raiz do repositório."
    )

if str(repo) not in sys.path:
    sys.path.insert(0, str(repo))

sys.dont_write_bytecode = True

print(f"Repositório: {repo}")
print(f"Pasta de testes: {repo / 'tests'}")
```

```python
import pytest

resultado = pytest.main([
    str(repo / "tests"),
    "-v",
    "-ra",
    "--tb=short",
    "-p", "no:cacheprovider"
])

assert resultado == 0, (
    f"Os testes falharam. Código de saída: {resultado}"
)

print("Todos os testes automatizados passaram.")
```

```python
resultado = pytest.main([
    str(repo / "tests" / "test_quality_rules.py"),
    "-v",
    "--tb=short",
    "-p", "no:cacheprovider"
])

assert resultado == 0, "Falha nos testes de deduplicação"
```
